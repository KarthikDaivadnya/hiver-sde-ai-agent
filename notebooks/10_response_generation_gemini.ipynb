````python
def build_judge_prompt(row):
    return f"""
You are evaluating an AI customer-support response.

Evaluate the generated response using ONLY the customer conversation,
the historical support evidence, the stated intent, and the stated
escalation decision provided below.

Do not assume policies, facts, or actions that are not supported by
the evidence.

CUSTOMER CONVERSATION:
{row["conversation_text"]}

INTENT:
{row["intent_label"]}

ESCALATION DECISION:
{row["policy_decision"]}

ESCALATION REASON:
{row["policy_reason_text"]}

HISTORICAL SUPPORT EVIDENCE:

EPISODE 1:
{row["retrieval_1_conversation_text"]}

EPISODE 2:
{row["retrieval_2_conversation_text"]}

EPISODE 3:
{row["retrieval_3_conversation_text"]}

GENERATED RESPONSE:
{row["groq_reply"]}

Score each dimension from 0 to 2.

CORRECTNESS:
0 = incorrect, misleading, or contradicts the evidence
1 = partially correct or materially incomplete
2 = correct and consistent with the evidence

RELEVANCE:
0 = does not address the customer's actual issue
1 = partially addresses the issue
2 = directly addresses the customer's actual issue

GROUNDING:
0 = contains unsupported or fabricated claims/actions
1 = mostly grounded but contains minor unsupported detail
2 = fully grounded in the supplied conversation/evidence

HELPFULNESS:
0 = provides no useful assistance
1 = somewhat useful
2 = provides an appropriate useful next step or response

TONE:
0 = inappropriate, unprofessional, or insensitive
1 = acceptable
2 = professional, concise, and appropriately empathetic

COMPLETENESS:
0 = misses the central support situation
1 = addresses it partially
2 = adequately handles the support situation given the available evidence

FAILURE FLAGS:
Select any applicable flags:
- FABRICATED_DETAIL
- UNSUPPORTED_ACTION
- WRONG_INTENT
- MISSED_ESCALATION
- UNNECESSARY_ESCALATION
- IRRELEVANT_RESPONSE
- INCOMPLETE_RESPONSE
- NONE

Return ONLY valid JSON in exactly this structure:

{{
  "correctness": 0,
  "relevance": 0,
  "grounding": 0,
  "helpfulness": 0,
  "tone": 0,
  "completeness": 0,
  "failure_flags": [],
  "brief_reason": ""
}}
"""
````

````python
import json
import re
import pandas as pd
from pathlib import Path

def parse_judge_json(raw_output):
    """
    Parse an LLM judge response into structured evaluation fields.
    Returns a dictionary with None values if parsing fails.
    """

    result = {
        "correctness": None,
        "relevance": None,
        "grounding": None,
        "helpfulness": None,
        "tone": None,
        "completeness": None,
        "failure_flags": None,
        "brief_reason": None,
        "parse_status": "FAILED"
    }

    if not isinstance(raw_output, str) or not raw_output.strip():
        return result

    text = raw_output.strip()

    # Remove Markdown code fences if the model ever returns them.
    text = re.sub(r"^```(?:json)?\s*", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\s*```$", "", text)

    # First attempt: parse the entire response.
    try:
        data = json.loads(text)
    except json.JSONDecodeError:

        # Second attempt: extract the first JSON object.
        match = re.search(r"\{.*\}", text, flags=re.DOTALL)

        if not match:
            return result

        try:
            data = json.loads(match.group(0))
        except json.JSONDecodeError:
            return result

    score_fields = [
        "correctness",
        "relevance",
        "grounding",
        "helpfulness",
        "tone",
        "completeness"
    ]

    valid = True

    for field in score_fields:
        value = data.get(field)

        if isinstance(value, bool):
            valid = False
            break

        try:
            value = int(value)
        except (TypeError, ValueError):
            valid = False
            break

        if value not in [0, 1, 2]:
            valid = False
            break

        result[field] = value

    if not valid:
        return result

    flags = data.get("failure_flags", [])

    if isinstance(flags, str):
        flags = [flags]

    if not isinstance(flags, list):
        return result

    result["failure_flags"] = flags
    result["brief_reason"] = str(data.get("brief_reason", "")).strip()
    result["parse_status"] = "SUCCESS"

    return result
````

````python
import time
import pandas as pd
from pathlib import Path

resume_output_path = Path(
    r"d:\BECAME_DEVELOPER\hiver-sde-ai-agent\data\processed\llm_judge_resume_results.csv"
)

def run_remaining_judges(
    pending_df,
    output_path=resume_output_path,
    delay_seconds=0.2
):
    """
    Resume-safe LLM-as-judge runner.

    Makes API calls only for pending rows.
    Saves every successful result immediately.
    Stops on rate-limit errors so progress is never wasted.
    """

    # Load previous resume results if they already exist.
    if output_path.exists():
        saved_results = pd.read_csv(output_path)
        completed_ids = set(saved_results["benchmark_id"].astype(str))
        print(f"Existing resume results: {len(saved_results)}")
    else:
        saved_results = pd.DataFrame()
        completed_ids = set()
        print("No previous resume results found.")

    pending = pending_df[
        ~pending_df["benchmark_id"].astype(str).isin(completed_ids)
    ].copy()

    print(f"Remaining API calls needed: {len(pending)}")

    for idx, (_, row) in enumerate(pending.iterrows(), start=1):

        benchmark_id = str(row["benchmark_id"])

        try:
            judge_prompt = build_judge_prompt(row)

            response = groq_client.chat.completions.create(
                model="openai/gpt-oss-20b",
                messages=[
                    {
                        "role": "user",
                        "content": judge_prompt
                    }
                ],
                temperature=0.0,
                max_completion_tokens=1024,
                reasoning_effort="low",
                stream=False
            )

            judge_output = response.choices[0].message.content

            new_result = pd.DataFrame([{
                "benchmark_id": benchmark_id,
                "judge_output": judge_output,
                "finish_reason": response.choices[0].finish_reason,
                "judge_status": "SUCCESS"
            }])

            saved_results = pd.concat(
                [saved_results, new_result],
                ignore_index=True
            )

            # Save immediately after every successful judgment.
            saved_results.to_csv(
                output_path,
                index=False,
                encoding="utf-8-sig"
            )

            print(
                f"[{idx}/{len(pending)}] "
                f"{benchmark_id} ✓"
            )

            time.sleep(delay_seconds)

        except Exception as e:

            error_text = str(e)

            print(
                f"\nSTOPPED at {benchmark_id}"
            )
            print(error_text)

            # Do not continue after a rate-limit error.
            if "429" in error_text or "rate_limit" in error_text.lower():
                print("\nRate limit detected.")
                print("Successful results have already been saved.")
                break

            # For another unexpected error, also stop rather than
            # silently skipping a benchmark.
            raise

    print("\n" + "=" * 60)
    print("RESUME RUN STATUS")
    print("=" * 60)
    print("Saved successful judgments:", len(saved_results))
    print("Output:", output_path)

    return saved_results
````

````python
import pandas as pd
import numpy as np
import ast
from pathlib import Path

# ---------------------------------------------------------
# 1. Load the complete 150 judge results
# ---------------------------------------------------------

checkpoint = pd.read_csv(
    r"d:\BECAME_DEVELOPER\hiver-sde-ai-agent\data\processed\llm_judge_checkpoint_28.csv"
)

resume = pd.read_csv(
    r"d:\BECAME_DEVELOPER\hiver-sde-ai-agent\data\processed\llm_judge_resume_results.csv"
)

judge_complete = pd.concat(
    [checkpoint, resume],
    ignore_index=True
)

judge_complete["benchmark_id"] = (
    judge_complete["benchmark_id"].astype(str).str.strip()
)

print("Judge results loaded:", len(judge_complete))
print("Unique IDs:", judge_complete["benchmark_id"].nunique())

# ---------------------------------------------------------
# 2. Parse judge outputs
# ---------------------------------------------------------

parsed_rows = []

for _, row in judge_complete.iterrows():
    parsed = parse_judge_json(row["judge_output"])
    parsed["benchmark_id"] = row["benchmark_id"]
    parsed_rows.append(parsed)

parsed_df = pd.DataFrame(parsed_rows)

# ---------------------------------------------------------
# 3. Merge with original benchmark
# ---------------------------------------------------------

evaluation_final = benchmark_df.merge(
    judge_complete[
        [
            "benchmark_id",
            "judge_output",
            "finish_reason",
            "judge_status"
        ]
    ],
    on="benchmark_id",
    how="inner",
    validate="one_to_one"
).merge(
    parsed_df,
    on="benchmark_id",
    how="inner",
    validate="one_to_one"
)

# ---------------------------------------------------------
# 4. Score columns
# ---------------------------------------------------------

score_columns = [
    "correctness",
    "relevance",
    "grounding",
    "helpfulness",
    "tone",
    "completeness"
]

for col in score_columns:
    evaluation_final[col] = pd.to_numeric(
        evaluation_final[col],
        errors="coerce"
    )

evaluation_final["overall_score"] = (
    evaluation_final[score_columns].mean(axis=1)
)

evaluation_final["overall_percentage"] = (
    evaluation_final["overall_score"] / 2.0 * 100
)

# ---------------------------------------------------------
# 5. Overall metrics
# ---------------------------------------------------------

overall_mean = evaluation_final[score_columns].mean()
overall_percentage = (overall_mean / 2.0) * 100

# ---------------------------------------------------------
# 6. Robust failure-flag normalization
# ---------------------------------------------------------

def normalize_failure_flags(value):

    # None / NaN
    if value is None:
        return []

    if isinstance(value, float) and np.isnan(value):
        return []

    # Already a list
    if isinstance(value, (list, tuple, set, np.ndarray)):
        result = []

        for item in value:
            if item is None:
                continue

            item = str(item).strip()

            if item and item != "NONE":
                result.append(item)

        return result

    # String representation of a list
    if isinstance(value, str):

        value = value.strip()

        if not value or value == "NONE":
            return []

        try:
            parsed = ast.literal_eval(value)

            if isinstance(parsed, (list, tuple, set)):
                return normalize_failure_flags(parsed)

        except Exception:
            pass

        return [value]

    return [str(value).strip()]


evaluation_final["failure_flags_normalized"] = (
    evaluation_final["failure_flags"]
    .apply(normalize_failure_flags)
)

# ---------------------------------------------------------
# 7. Failure-mode counts
# ---------------------------------------------------------

failure_counts = {}

for flags in evaluation_final["failure_flags_normalized"]:

    for flag in flags:

        if flag == "NONE" or not flag:
            continue

        failure_counts[flag] = (
            failure_counts.get(flag, 0) + 1
        )

failure_table = pd.DataFrame(
    [
        {
            "failure_mode": flag,
            "count": count,
            "percentage_of_150": count / len(evaluation_final) * 100
        }
        for flag, count in failure_counts.items()
    ]
)

if not failure_table.empty:
    failure_table = failure_table.sort_values(
        ["count", "failure_mode"],
        ascending=[False, True]
    ).reset_index(drop=True)

# ---------------------------------------------------------
# 8. Per-intent evaluation
# ---------------------------------------------------------

per_intent = (
    evaluation_final
    .groupby("intent_label")
    .agg(
        examples=("benchmark_id", "count"),
        overall_score=("overall_score", "mean"),
        correctness=("correctness", "mean"),
        relevance=("relevance", "mean"),
        grounding=("grounding", "mean"),
        helpfulness=("helpfulness", "mean"),
        tone=("tone", "mean"),
        completeness=("completeness", "mean")
    )
    .reset_index()
)

per_intent["overall_percentage"] = (
    per_intent["overall_score"] / 2.0 * 100
)

# ---------------------------------------------------------
# 9. By escalation policy
# ---------------------------------------------------------

by_policy = (
    evaluation_final
    .groupby("policy_decision")
    .agg(
        examples=("benchmark_id", "count"),
        overall_score=("overall_score", "mean"),
        correctness=("correctness", "mean"),
        relevance=("relevance", "mean"),
        grounding=("grounding", "mean"),
        helpfulness=("helpfulness", "mean"),
        tone=("tone", "mean"),
        completeness=("completeness", "mean")
    )
    .reset_index()
)

by_policy["overall_percentage"] = (
    by_policy["overall_score"] / 2.0 * 100
)

# ---------------------------------------------------------
# 10. Save final artifacts
# ---------------------------------------------------------

output_dir = Path(
    r"d:\BECAME_DEVELOPER\hiver-sde-ai-agent\data\processed"
)

evaluation_final.to_csv(
    output_dir / "llm_judge_evaluation_final_150.csv",
    index=False,
    encoding="utf-8-sig"
)

failure_table.to_csv(
    output_dir / "llm_judge_failure_modes_final.csv",
    index=False,
    encoding="utf-8-sig"
)

per_intent.to_csv(
    output_dir / "llm_judge_per_intent_final.csv",
    index=False,
    encoding="utf-8-sig"
)

by_policy.to_csv(
    output_dir / "llm_judge_by_policy_final.csv",
    index=False,
    encoding="utf-8-sig"
)

# ---------------------------------------------------------
# 11. Final output
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL LLM-AS-JUDGE RESULTS — 150 EXAMPLES")
print("=" * 70)

print(f"\nBenchmark examples: {len(evaluation_final)}")

print("\nOverall dimension scores:")

for col in score_columns:
    print(
        f"{col:15s}: "
        f"{overall_mean[col]:.3f}/2 "
        f"({overall_percentage[col]:.1f}%)"
    )

print(
    f"\nOverall mean: "
    f"{evaluation_final['overall_score'].mean():.3f}/2 "
    f"({evaluation_final['overall_percentage'].mean():.1f}%)"
)

print("\nFailure modes:")

if failure_table.empty:
    print("No failure flags recorded.")
else:
    print(
        failure_table.to_string(index=False)
    )

print("\nPer-policy results:")
print(
    by_policy.round(3).to_string(index=False)
)

print("\nFiles saved:")
print(output_dir / "llm_judge_evaluation_final_150.csv")
print(output_dir / "llm_judge_failure_modes_final.csv")
print(output_dir / "llm_judge_per_intent_final.csv")
print(output_dir / "llm_judge_by_policy_final.csv")

print("=" * 70)
````
